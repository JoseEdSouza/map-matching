```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import seaborn as sns
```

```python
DATA_BASE_PATH = Path("./data").resolve().absolute()
```

# Milan

```python
milan_df = pd.read_csv(DATA_BASE_PATH / "milan_fcd_prep.csv")
milan_df.head()
```

```python
milan_df.shape
```

```python
milan_df.iloc[:, 1:].max().max()
```

```python
milan_df.iloc[:, 1:].min().min()
```

```python
print(milan_df.iloc[0, 1::2])
print(milan_df.iloc[0, 2::2])
```

```python
milan_df.shape[0] // 50
```

```python
plt.figure(figsize=(9, 9))

for i in range(0, milan_df.shape[0], 50):
    plt.plot(milan_df.iloc[i, 1::2], milan_df.iloc[i, 2::2], "m", alpha=0.05)
```

```python
fig, ax = plt.subplots(figsize=(16, 16))

s = ax.scatter([], [], color="m", alpha=0.05)
ax.set_xlim(-1, 1)
ax.set_ylim(-1, 1)
s.set_animated(True)


def update(i):
    i -= 8
    x_col = f"x(t{i:+})" if i != 0 else "x(t)"
    y_col = f"y(t{i:+})" if i != 0 else "y(t)"
    s.set_offsets(milan_df[[x_col, y_col]])
    # fcd = ax.scatter(milan_df[x_col], milan_df[y_col], alpha=0.05)
    # plt.pause(1)
    # fcd.set_offsets(np.matrix.transpose([milan_df[x_col], milan_df[y_col]]))


ani = animation.FuncAnimation(fig, update, frames=20)

# plt.show()

ani.save("milan_fcd.gif", writer="pillow")
```
