# Introdução
Este notebook consolida a análise comparativa das ferramentas de map matching no modo online, avaliando como a taxa de amostragem das trajetórias afeta acurácia, latência, custo computacional e throughput.

A análise usa os CSVs agregados em `results/` (métricas E2E, match e Prometheus) e produz tabelas e figuras para responder às perguntas de pesquisa (QP1–QP4). As figuras são salvas em `OUTPUT_PATH`, mantendo rastreabilidade dos resultados.

Fluxo de trabalho:
- carregar os dados agregados e validar consistência;
- extrair `sample_rate` a partir do `dataset_id`;
- agregar métricas por ferramenta e taxa de amostragem;
- gerar gráficos e tabelas para apoiar as conclusões.

Notas de leitura:
- qualidade (F1, error_rate, Newson–Krumm) melhora com F1 alto e erros baixos;
- latência, CPU e memória são melhores quanto menores;
- throughput é melhor quanto maior.

## Setup
Aqui concentramos todos os imports e constantes (paths, cores e ordens), além das funções auxiliares de leitura, agregação e plotagem. Isso reduz duplicação e garante consistência entre as seções.


```python
from pathlib import Path
import re

import duckdb
import numpy as np
import pandas as pd
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
RESULTS_PATH = Path.cwd() / "results"
OUTPUT_PATH = RESULTS_PATH / "figures"
OUTPUT_PATH.mkdir(parents=True, exist_ok=True)

E2E_PATH = RESULTS_PATH / "aggregated_e2e_metrics.csv"
MATCH_PATH = RESULTS_PATH / "aggregated_match_metrics.csv"
PROMETHEUS_PATH = RESULTS_PATH / "aggregated_prometheus_metrics.csv"
FINAL_PATH = RESULTS_PATH / "final_runs_table.csv"

SAMPLE_RATE_COLUMNS = ("dataset_id_mm", "dataset_id_prom")
SAMPLE_RATE_PROM_COLUMN = "dataset_id_prom"
LATENCY_COL = "e2e_avg_step_latency_ms"
E2E_TIME_COL = "duration_s"
EPS = 1e-9

TOOL_COLORS = {
    "barefoot": "tab:blue",
    "graphhopper": "tab:orange",
    "graphium": "tab:green",
    "osrm": "tab:red",
}
TOOL_ORDER = ["barefoot", "graphhopper", "graphium", "osrm"]

VARIANT_COLORS = {
    "barefoot": "tab:blue",
    "graphhopper": "tab:orange",
    "graphium (cold)": "tab:green",
    "graphium (warm)": "tab:purple",
    "osrm": "tab:red",
}
VARIANT_ORDER = ["barefoot", "graphhopper", "graphium (cold)", "graphium (warm)", "osrm"]

DEFAULT_FIGSIZE = (8, 5)
```

```python
def read_csv(path: Path) -> pd.DataFrame:
    return duckdb.read_csv(path).to_df()


def add_throughput_km_per_s(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out["e2e_throughput_km_per_s"] = out["total_gt_length"] / out["e2e_execution_time_s"]
    return out


def extract_sample_rate(dataset_id: str) -> float | None:
    match = re.search(r"sr_(\d+(\.\d+)?)", str(dataset_id))
    return float(match.group(1)) if match else None


def add_sample_rate(df: pd.DataFrame, column: str) -> pd.DataFrame:
    out = df.copy()
    out["sample_rate"] = out[column].apply(extract_sample_rate)
    return out


def add_sample_rate_from_columns(df: pd.DataFrame, columns: tuple[str, ...]) -> pd.DataFrame:
    for column in columns:
        if column in df.columns:
            return add_sample_rate(df, column)
    out = df.copy()
    out["sample_rate"] = np.nan
    return out


def resolve_output_path(file_name: str | Path | None) -> Path | None:
    if not file_name:
        return None
    file_path = Path(file_name)
    return file_path if file_path.is_absolute() else OUTPUT_PATH / file_path


def save_figure(file_name: str | Path | None) -> None:
    out_path = resolve_output_path(file_name)
    if out_path:
        plt.savefig(out_path)


def plot_line_by_group(
    df: pd.DataFrame,
    x: str,
    y: str,
    group: str,
    *,
    title: str | None = None,
    xlabel: str | None = None,
    ylabel: str | None = None,
    figsize: tuple[int, int] | None = None,
    order: list[str] | None = None,
    colors: dict[str, str] | None = None,
    file_name: str | None = None,
) -> None:
    if figsize:
        plt.figure(figsize=figsize)
    else:
        plt.figure(figsize=DEFAULT_FIGSIZE)

    palette = colors or TOOL_COLORS
    groups = order if order is not None else df[group].dropna().unique()
    for name in groups:
        g = df[df[group] == name]
        if g.empty:
            continue
        plt.plot(
            g[x],
            g[y],
            marker="o",
            label=name,
            color=palette.get(name),
        )

    plt.xlabel(xlabel or x)
    plt.ylabel(ylabel or y)
    if title:
        plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    save_figure(file_name)
    plt.show()


def shade_color(color: str, amount: float = 0.5) -> str:
    try:
        rgb = np.array(mcolors.to_rgb(color))
        rgb = rgb * (1 - amount)
        return mcolors.to_hex(rgb)
    except ValueError:
        return color


def plot_grouped_bars_by_batch(
    agg_df: pd.DataFrame,
    value_col: str,
    ylabel: str,
    matcher_name_col: str = "matcher_name",
    title: str | None = None,
    batch_col: str = "batch_size_mm",
    log_scale: bool = False,
    file_name: str | None = None,
    order: list[str] | None = None,
    colors: dict[str, str] | None = None,
) -> None:
    bvals = sorted(agg_df[batch_col].unique())
    n_b = len(bvals)

    dark_levels = np.linspace(0.10, 0.70, n_b) if n_b > 1 else [0.0]
    b_to_dark = {b: dark_levels[i] for i, b in enumerate(bvals)}

    palette = colors or TOOL_COLORS
    order = order or sorted(agg_df[matcher_name_col].unique())

    x = np.arange(len(order))
    total_width = 0.8
    bar_w = total_width / max(n_b, 1)

    plt.figure(figsize=DEFAULT_FIGSIZE)

    for i, b in enumerate(bvals):
        vals = []
        for m in order:
            row = agg_df[(agg_df[matcher_name_col] == m) & (agg_df[batch_col] == b)]
            vals.append(float(row[value_col].iloc[0]) if not row.empty else 0.0)

        offset = -total_width / 2 + (i + 0.5) * bar_w
        bar_colors = [shade_color(palette.get(m, "gray"), b_to_dark[b]) for m in order]

        plt.bar(
            x + offset,
            vals,
            width=bar_w,
            color=bar_colors,
            edgecolor="black",
            linewidth=0.3,
        )

    if log_scale:
        plt.yscale("log")

    plt.xticks(x, order)
    plt.xlabel("Tool")
    plt.ylabel(ylabel)
    if title:
        plt.title(title)
    plt.grid(True, axis="y", linestyle="--", alpha=0.7)

    plt.xlim(-0.6, len(order) - 0.4 + 1.0)

    legend_handles = []
    for i, b in enumerate(bvals):
        gray_val = 1.0 - dark_levels[i]
        gray_color = (gray_val, gray_val, gray_val)
        batch_label = f"{int(b) if float(b).is_integer() else b}"
        patch = mpatches.Patch(
            facecolor=gray_color,
            edgecolor="black",
            linewidth=0.5,
            label=f"batch={batch_label}",
        )
        legend_handles.append(patch)

    plt.legend(
        handles=legend_handles,
        title="Batch size",
        loc="center right",
        bbox_to_anchor=(0.98, 0.5),
    )

    plt.tight_layout()
    save_figure(file_name)
    plt.show()
```

```python
def pct95(series: pd.Series) -> float:
    return np.percentile(series, 95)


def get_length_column(df: pd.DataFrame) -> str | None:
    if "total_gt_length" in df.columns:
        return "total_gt_length"
    if "total_added_length" in df.columns:
        return "total_added_length"
    return None


def build_points_per_run(e2e_df: pd.DataFrame) -> pd.DataFrame:
    e2e_pts = e2e_df.copy()
    e2e_pts["cum_points"] = pd.to_numeric(e2e_pts["cum_points"], errors="coerce")
    e2e_pts = e2e_pts.dropna(subset=["run_id", "cum_points"])
    return e2e_pts.groupby("run_id", as_index=False).agg(total_points=("cum_points", "max"))


def build_throughput_runs(
    final_df: pd.DataFrame, e2e_df: pd.DataFrame
) -> tuple[pd.DataFrame, str | None, str | None]:
    length_col = get_length_column(final_df)
    batch_col = None
    if "batch_size_mm" in final_df.columns:
        batch_col = "batch_size_mm"
    elif "batch_size_prom" in final_df.columns:
        batch_col = "batch_size_prom"

    df = final_df.copy()
    df["duration_s"] = pd.to_numeric(df["duration_s"], errors="coerce")
    df = df.dropna(subset=["run_id", "matcher_name", "duration_s"])
    df = df[df["duration_s"] > 0]

    if length_col:
        df[length_col] = pd.to_numeric(df[length_col], errors="coerce")

    points_per_run = build_points_per_run(e2e_df)
    df = df.merge(points_per_run, on="run_id", how="left")
    df = df.dropna(subset=["total_points"])
    df["total_points"] = pd.to_numeric(df["total_points"], errors="coerce")
    df = df[df["total_points"] > 0]

    df["throughput_pps_calc"] = df["total_points"] / df["duration_s"]
    if length_col:
        df["throughput_kmps_calc"] = (df[length_col] / 1000.0) / df["duration_s"]

    return df, length_col, batch_col
```

Nesta seção avaliamos a qualidade do map matching no modo online, comparando F1-score, error_rate e Newson–Krumm por ferramenta e por taxa de amostragem.

Metodologia:

**QP1.** Melhor acurácia no online por taxa de amostragem (média de `accuracy` por ferramenta):

```python
final_runs_df_agg_full = duckdb.query(
    """
    SELECT
        matcher_name,
        mode,
        adapter_prom AS adapter,
        AVG(e2e_throughput_pps) AS e2e_throughput_pps_avg,
        AVG(e2e_avg_step_latency_ms) AS e2e_avg_step_latency_ms_avg,
        AVG(e2e_p95_step_latency_ms) AS e2e_p95_step_latency_ms_avg,
        AVG(e2e_p99_step_latency_ms) AS e2e_p99_step_latency_ms_avg,
        SUM(e2e_execution_time_s) AS e2e_execution_time_s_sum,
        AVG(cpu_avg) AS sr_cpu_avg,
        AVG(mem_avg_mb) AS sr_mem_mb_avg,
        AVG(accuracy) AS match_accuracy_avg,
        AVG(precision) AS match_precision_avg,
        AVG(recall) AS match_recall_avg,
        AVG(f1_score) AS match_f1_score_avg,
        AVG(error_rate) AS match_error_rate_avg,
        AVG(newson_krumm_error) AS match_newson_krumm_error_avg,
        SUM(total_gt_length) / AVG(e2e_execution_time_s) AS e2e_throughput_km_per_s
    FROM final_runs_df
    GROUP BY matcher_name, mode, adapter
    ORDER BY matcher_name, mode
    """
).to_df()
final_runs_df_agg_full
```

```python
final_runs_df_agg_summary = (
    final_runs_df.groupby(["matcher_name", "mode"], as_index=False)
    .agg(
        e2e_throughput_pps=("e2e_throughput_pps", "mean"),
        e2e_avg_step_latency_ms=("e2e_avg_step_latency_ms", "mean"),
        e2e_p95_step_latency_ms=("e2e_p95_step_latency_ms", "mean"),
        e2e_p99_step_latency_ms=("e2e_p99_step_latency_ms", "mean"),
        cpu_avg=("cpu_avg", "mean"),
        mem_avg_mb=("mem_avg_mb", "mean"),
        accuracy=("accuracy", "mean"),
    )
)
final_runs_df_agg_summary
```

## QP2 — Eficiência temporal (latência)
Aqui medimos o tempo de processamento no modo online sob duas visões complementares:
- latência média por etapa (step latency), agregada por ferramenta e `sample_rate`;
- tempo total por execução (`duration_s`), que captura o custo completo do processamento.

Para o Graphium, separamos cenários **cold** e **warm** para isolar o impacto do primeiro passo.


```python
# --------------------------------------------------
# Load data
# --------------------------------------------------
df = add_sample_rate_from_columns(final_runs_df, (SAMPLE_RATE_PROM_COLUMN,))

# --------------------------------------------------
# Extract sample rate from dataset_id_prom
# Example: ohare_filtered_vid_1139_sr_5.0 -> 5.0
# --------------------------------------------------
df = df.dropna(subset=["sample_rate"]).copy()
df["sample_rate"] = df["sample_rate"].astype(float)

# --------------------------------------------------
# QP2: Temporal efficiency (latency) vs sample rate
# Choose one latency metric to be consistent.
# Recommended: e2e_avg_step_latency_ms (average step latency in ms)
# --------------------------------------------------
lat_col = LATENCY_COL

# Keep only rows with latency available (some rows may be missing, e.g., failed runs)
df_lat = df.dropna(subset=[lat_col]).copy()

qp2_agg = (
    df_lat.groupby(["matcher_name", "sample_rate"], as_index=False)
    .agg(
        latency_mean_ms=(lat_col, "mean"),
        runs=(lat_col, "count"),
    )
    .sort_values(["matcher_name", "sample_rate"])
)

print(qp2_agg)

plot_line_by_group(
    qp2_agg,
    x="sample_rate",
    y="latency_mean_ms",
    group="matcher_name",
    xlabel="Sample Rate (s)",
    ylabel="Average step latency (ms)",
    title="Average step latency vs Sample Rate",
    figsize=DEFAULT_FIGSIZE,
    order=TOOL_ORDER,
    colors=TOOL_COLORS,
    file_name="qp2_latency_vs_sample_rate.pdf",
)
```

```python
# --------------------------------------------------
# Load datasets
# --------------------------------------------------
final_runs_with_sr = add_sample_rate_from_columns(final_runs_df, SAMPLE_RATE_COLUMNS)
e2e_steps_df = e2e_df.copy()

# --------------------------------------------------
# Join E2E steps with run-level metadata
# --------------------------------------------------
meta_cols = ["run_id", "matcher_name_mm", "sample_rate"]
for c in ["batch_size_mm", "dataset_id_mm"]:
    if c in final_runs_with_sr.columns:
        meta_cols.append(c)

df = e2e_steps_df.merge(final_runs_with_sr[meta_cols], on="run_id", how="left")

# Keep only rows we can analyze
df = df.dropna(subset=["matcher_name", "sample_rate", "step_latency_ms"]).copy()

# Ensure numeric
df["sample_rate"] = df["sample_rate"].astype(float)

# --------------------------------------------------
# Build matcher_variant with Graphium cold/warm
# Warm rule:
# - For runs with >1 step: drop step_index == 0
# - For runs with exactly 1 step: keep step_index == 0 (so warm exists)
# --------------------------------------------------
non_graph = df[df["matcher_name"] != "graphium"].copy()
non_graph["matcher_variant"] = non_graph["matcher_name"]

graph = df[df["matcher_name"] == "graphium"].copy()

# Cold = all steps
graph_cold = graph.copy()
graph_cold["matcher_variant"] = "graphium (cold)"

# Determine runs that have only 1 step
# (step_count == 1) OR (max step_index == 0) - both are safe checks
graph_run_steps = (
    graph.groupby("run_id", as_index=False)
    .agg(step_count=("step_index", "count"), max_step=("step_index", "max"))
)
graph = graph.merge(graph_run_steps, on="run_id", how="left")

# Warm selection:
# - if run has only 1 step (max_step == 0 or step_count == 1): keep it
# - else: exclude step 0
warm_mask = (
    (graph["step_count"] == 1)
    | (graph["max_step"] == 0)
    | (graph["step_index"] != 0)
)
graph_warm = graph[warm_mask].copy()
graph_warm["matcher_variant"] = "graphium (warm)"

# Combine dataset for analysis
df_all = pd.concat([non_graph, graph_cold, graph_warm], ignore_index=True)

# --------------------------------------------------
# QP2 aggregation: latency vs sample_rate
# --------------------------------------------------
qp2_sr = (
    df_all.groupby(["matcher_variant", "sample_rate"], as_index=False)
    .agg(
        latency_mean_ms=("step_latency_ms", "mean"),
        steps=("step_latency_ms", "count"),
        runs=("run_id", pd.Series.nunique),
    )
    .sort_values(["matcher_variant", "sample_rate"])
)

print(qp2_sr)

# Plot: latency mean vs sample rate
plot_line_by_group(
    qp2_sr,
    x="sample_rate",
    y="latency_mean_ms",
    group="matcher_variant",
    xlabel="Sample rate (s)",
    ylabel="Average step latency (ms)",
    order=VARIANT_ORDER,
    colors=VARIANT_COLORS,
    file_name="qp2_latency_vs_sample_rate.pdf",
)

# --------------------------------------------------
# QP2 aggregation: latency vs batch_size (if available)
# --------------------------------------------------
batch_col = "batch_size_mm" if "batch_size_mm" in df_all.columns else None

if batch_col:
    df_b = df_all.dropna(subset=[batch_col]).copy()
    df_b[batch_col] = pd.to_numeric(df_b[batch_col], errors="coerce")
    df_b = df_b.dropna(subset=[batch_col])

    qp2_bs = (
        df_b.groupby(["matcher_variant", batch_col], as_index=False)
        .agg(
            latency_mean_ms=("step_latency_ms", "mean"),
            steps=("step_latency_ms", "count"),
            runs=("run_id", pd.Series.nunique),
        )
        .sort_values(["matcher_variant", batch_col])
    )

    print(qp2_bs)

    plot_grouped_bars_by_batch(
        qp2_bs,
        value_col="latency_mean_ms",
        ylabel="Average step latency (ms)",
        matcher_name_col="matcher_variant",
        file_name="qp2_latency_vs_batch_size.pdf",
        colors=VARIANT_COLORS,
        order=[o for o in VARIANT_ORDER if o != "barefoot"],
    )
else:
    print("batch_size_mm não encontrado no metadata; pulei o gráfico por batch size.")
```

## QP3 — Eficiência computacional (custo vs qualidade)
Relacionamos qualidade (F1, error_rate) com consumo de CPU e memória. A ideia é observar trade-offs claros entre precisão e custo computacional.

Metodologia:
- agregar métricas por ferramenta × taxa de amostragem;
- calcular razões simples (cpu_per_f1 e mem_per_f1);
- gerar gráficos de custo e um scatter de trade-off qualidade vs CPU.


```python
# --------------------------------------------------
# Preparar dados (run-level)
# --------------------------------------------------
df = add_sample_rate_from_columns(final_runs_df, SAMPLE_RATE_COLUMNS)

df = df.dropna(subset=["sample_rate", "matcher_name"]).copy()
df["sample_rate"] = df["sample_rate"].astype(float)

# Se quiser focar só no online, descomente:
# df = df[df["mode"].eq("online")].copy()

# --------------------------------------------------
# Agregação QP3: custo + qualidade por ferramenta × sample_rate
# --------------------------------------------------
qp3 = (
    df.groupby(["matcher_name", "sample_rate"], as_index=False)
    .agg(
        # Qualidade
        f1_mean=("f1_score", "mean"),
        err_rate_mean=("error_rate", "mean"),
        nk_err_mean=("newson_krumm_error", "mean"),

        # Custo
        cpu_avg_mean=("cpu_avg", "mean"),
        mem_avg_mean=("mem_avg_mb", "mean"),

        # Opcional
        cpu_total_core_s_mean=("cpu_total_core_s", "mean"),
        mem_peak_mean=("mem_peak_mb", "mean"),

        runs=("run_id", "count"),
    )
    .sort_values(["matcher_name", "sample_rate"])
)

print(qp3)

# --------------------------------------------------
# Métricas simples de custo por qualidade
# --------------------------------------------------
qp3["cpu_per_f1"] = qp3["cpu_avg_mean"] / (qp3["f1_mean"] + EPS)
qp3["mem_per_f1"] = qp3["mem_avg_mean"] / (qp3["f1_mean"] + EPS)

# --------------------------------------------------
# Plot 1: CPU avg vs sample_rate (linhas)
# --------------------------------------------------
plot_line_by_group(
    qp3,
    x="sample_rate",
    y="cpu_avg_mean",
    group="matcher_name",
    xlabel="Sample rate (s)",
    ylabel="Average CPU (cpu_avg)",
    order=TOOL_ORDER,
    colors=TOOL_COLORS,
    file_name="qp3_cpu_vs_sample_rate.pdf",
)

# --------------------------------------------------
# Plot 2: Mem avg vs sample_rate (linhas)
# --------------------------------------------------
plot_line_by_group(
    qp3,
    x="sample_rate",
    y="mem_avg_mean",
    group="matcher_name",
    xlabel="Sample rate (s)",
    ylabel="Average memory (MB)",
    order=TOOL_ORDER,
    colors=TOOL_COLORS,
    file_name="qp3_mem_vs_sample_rate.pdf",
)

# --------------------------------------------------
# Plot 3: Qualidade vs CPU (trade-off)
# (cada ponto = uma taxa de amostragem)
# --------------------------------------------------
plt.figure(figsize=DEFAULT_FIGSIZE)
annotated_special = False
for name in TOOL_ORDER:
    g = qp3[qp3["matcher_name"] == name]
    if g.empty:
        continue
    plt.scatter(g["cpu_avg_mean"], g["f1_mean"], label=name, color=TOOL_COLORS.get(name))
    for _, row in g.iterrows():
        plt.annotate(
            f"{int(row['sample_rate'])}",
            (row["cpu_avg_mean"], row["f1_mean"]),
            fontsize=7,
            xytext=(4, 0.5),
            textcoords="offset points",
        )
        if (not annotated_special) and ("target_sr" in globals()) and (row["sample_rate"] == target_sr):
            plt.annotate(
                f"taxa de amostragem = {int(row['sample_rate'])} s",
                xy=(row["cpu_avg_mean"], row["f1_mean"]),
                xytext=(30, 30),
                textcoords="offset points",
                arrowprops=dict(arrowstyle="->", color=TOOL_COLORS.get(name)),
                bbox=dict(boxstyle="round,pad=0.3", fc="w", alpha=0.9),
                fontsize=9,
                color=TOOL_COLORS.get(name),
            )
            annotated_special = True

plt.xlabel("Average CPU (cpu_avg)")
plt.ylabel("Average F1-score")
plt.legend()
plt.grid(True)
plt.tight_layout()
save_figure("qp3_tradeoff_quality_cpu.svg")
save_figure("qp3_tradeoff_quality_cpu.pdf")
plt.show()
```

## QP4 — Throughput (capacidade de processamento)
Quantificamos a vazão em pontos/s e, quando possível, em km/s. O throughput é calculado a partir de `cum_points` (E2E) e `duration_s`, agregado por ferramenta e `sample_rate`.

Também avaliamos a influência do batch size quando a informação está disponível.


```python
# -----------------------------
# 1) sample_rate no final_runs_df
# -----------------------------
final_runs_with_sr = add_sample_rate_from_columns(final_runs_df, SAMPLE_RATE_COLUMNS)

throughput_df, length_col, batch_col = build_throughput_runs(final_runs_with_sr, e2e_df)

qp4_df = throughput_df.dropna(subset=["sample_rate"]).copy()
qp4_df["sample_rate"] = qp4_df["sample_rate"].astype(float)

# -----------------------------
# 2) Agregar por ferramenta × sample_rate (inclui p50 e p95)
# -----------------------------
agg_cols = {
    "throughput_pps_mean": ("throughput_pps_calc", "mean"),
    "throughput_pps_p50": ("throughput_pps_calc", "median"),
    "throughput_pps_p95": ("throughput_pps_calc", pct95),
    "runs": ("run_id", "count"),
}
if length_col:
    agg_cols["throughput_kmps_mean"] = ("throughput_kmps_calc", "mean")
    agg_cols["throughput_kmps_p50"] = ("throughput_kmps_calc", "median")
    agg_cols["throughput_kmps_p95"] = ("throughput_kmps_calc", pct95)

qp4_calc = (
    qp4_df.groupby(["matcher_name", "sample_rate"], as_index=False)
    .agg(**agg_cols)
    .sort_values(["matcher_name", "sample_rate"])
)

print(qp4_calc)

# -----------------------------
# 3) Plot: linhas por sample_rate (pps)
# -----------------------------
plot_line_by_group(
    qp4_calc,
    x="sample_rate",
    y="throughput_pps_p95",
    group="matcher_name",
    xlabel="Sample rate (s)",
    ylabel="p95 throughput (points/s)",
    file_name="qp4_throughput_pps_by_sample_rate.pdf",
    colors=TOOL_COLORS,
    order=TOOL_ORDER,
)

# -----------------------------
# 4) Plot: linhas por sample_rate (km/s), se houver length
# -----------------------------
if length_col:
    plot_line_by_group(
        qp4_calc,
        x="sample_rate",
        y="throughput_kmps_p95",
        group="matcher_name",
        xlabel="Sample rate (s)",
        ylabel="p95 throughput (km/s)",
        file_name="qp4_throughput_kmps_by_sample_rate.pdf",
        colors=TOOL_COLORS,
        order=TOOL_ORDER,
    )
else:
    print("Sem coluna de comprimento (total_gt_length/total_added_length). Pulei km/s.")
```

# Conclusão
Os resultados indicam um padrão consistente: **barefoot** se destaca em acurácia e latência no modo online, enquanto **graphium** tende a entregar maior throughput em taxas de amostragem menores. O trade-off custo–qualidade (CPU/memória vs F1) reforça a vantagem do barefoot quando o objetivo é obter precisão com baixo custo computacional.

Em cenários com alta demanda de vazão ou processamento em lote, graphium pode ser competitivo, especialmente quando o aquecimento (warm) é considerado. A escolha final depende do equilíbrio desejado entre precisão, tempo de resposta e capacidade de processamento.
